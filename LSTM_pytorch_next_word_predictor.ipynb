```python
sentences = []
```

```python
path = "/content/1661-0.txt"

with open(path) as txt:
  for i in txt:
    sentences.append(i.strip())
```

```python
def tokenize(text):
  text = text.lower()
  text = text.replace("?" , "")
  text = text.replace("'" , "")
  return text.split()
```

```python
vocal = {"<unk>":0}
```

```python
for i in sentences:
  x = tokenize(i)
  for j in x:
    if j in vocal:
      pass
    else:
      vocal[j] = len(vocal)
```

```python
print(vocal)
```

```python
# extract sentences
sentences.remove("")
sentences.remove('\ufeff')
```

```python
print(sentences[0])
```

```python
def text_to_index(text):
  indexed_text = []
  for j in tokenize(text):
      if j in vocal:
        indexed_text.append(vocal[j])
      else :
        indexed_text.append(vocal["<unk>"])
  return indexed_text
```

```python
text_to_index("hello how are you ")
```

```python
def sentence_to_data(x):
  sentences = []
  for i in x:
    indexed_text = text_to_index(i)
    sentences.append(indexed_text)
  return sentences
```

```python
number_sentence = sentence_to_data(sentences)
print(sentences[0])
number_sentence[0]
```

```python
training_sequence = []
for sen in number_sentence:
  for i in range(1 , len(sen)):
    training_sequence.append(sen[:i+1])
```
